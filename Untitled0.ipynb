#           去重、集合運算比較參加者名單，並計算各場人數、報名人次與應收費用。

```python
# ==============================================================================
# 【測試一：原始資料】
# ==============================================================================
print("==================== 【測試一：原始資料】 ====================")

# 1. 準備基礎資料 (tuple, str, dict)
activity_names = ("桌遊體驗", "電影欣賞")  # 使用 tuple 保存固定的兩個活動名稱
board_raw = " Amy, Ben, amy, Cora "        # 桌遊原始名單字串
movie_raw = " ben, Dora, ERIC, ben "       # 電影原始名單字串
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}  # 使用 dict 記錄各活動每人費用

# 2. 讀取活動名稱，並整理字串為保留重複項目的 list
# 使用索引從 tuple 取出活動名稱
act_board = activity_names[0]
act_movie = activity_names[1]

# 先用 replace(" ", "") 移除所有半形空白，再用 lower() 統一轉成小寫，最後用 split(",") 拆成 list
# 注意：保留原始 board_raw 與 movie_raw 不覆蓋，且此階段保留重複項目與原本順序
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 3. 使用集合 (set) 去重並進行名單比較
board_set = set(board_list)  # 轉成 set 自動去除同一場的重複報名者
movie_set = set(movie_list)

both_set = board_set & movie_set          # 交集 (&)：兩場都有參加的共同參加者
all_set = board_set | movie_set           # 聯集 (|)：至少參加一場的全部參加者
only_board_set = board_set - movie_set    # 差集 (-)：只參加桌遊、沒參加電影者
only_movie_set = movie_set - board_set    # 差集 (-)：只參加電影、沒參加桌遊者

# 4. 排序名單並使用切片取得通知名單前兩位
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

# 使用切片 [:2] 取得全部參加者排序後的前兩位作為預覽
notify_preview = all_sorted[:2]

# 5. 建立摘要字典 (dict)，計算人數、人次與應收費用
# 依照去重後的人數 (len) 與 unit_fees 計算費用，字典內保留純數字
board_count = len(board_set)
movie_count = len(movie_set)
board_fee_total = board_count * unit_fees[act_board]
movie_fee_total = movie_count * unit_fees[act_movie]

summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": board_count + movie_count,
    "桌遊應收": board_fee_total,
    "電影應收": movie_fee_total,
    "總應收": board_fee_total + movie_fee_total,
}

# 6. 使用 f-string 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元\n")
```
